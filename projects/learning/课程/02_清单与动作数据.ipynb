分类器处理一段动作，不是根据文件夹名字猜答案。一个片段有多帧Depth_Color与IR。sample的index是样本位置，label是类别编号；它们不相等。训练集用于学参数，验证集用于评估和选择方案。

## 数据规模与类别映射


```python
from collections import Counter
for split in ['train', 'validation']:
    data = L.manifest(split)
    counts = Counter(row['label'] for row in data['samples'])
    print(split, '片段数', len(data['samples']), '类别数', len(data['classes']), '覆盖类别数', len(counts))
```

## 核对时间位置


```python
from depth_ir_io import sample_frame_lists, synchronized_paths, uniform_indices
path, row, classes = L.sample_path('train', 0)
depth, ir = sample_frame_lists(path)
synchronized_paths(depth, ir)
print('两种模态帧数：', len(depth), len(ir))
print('取6帧的位置：', uniform_indices(len(depth), 6))
print('真实类别：', row['label'], classes[row['label']])
```
