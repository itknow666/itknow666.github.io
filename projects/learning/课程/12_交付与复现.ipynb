## 记录明确的权重身份


```python
import hashlib
run = L.PROJECT / L.CONFIG['reference_run']
config = L.read(run / 'inference-config.json')
weight_path = run / config['weights_file']
digest = hashlib.sha256()
with weight_path.open('rb') as stream:
    for block in iter(lambda: stream.read(4*1024*1024), b''):
        digest.update(block)
print('weights_file', config['weights_file'])
print('SHA256', digest.hexdigest())
print('报告里写清训练配置与权重身份，不只写“用了最新模型”。')
```
