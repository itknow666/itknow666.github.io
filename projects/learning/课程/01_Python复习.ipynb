## 从磁盘到Python对象


```python
import json
path = L.PROJECT / 'data/raw/train/manifest.json'
data = json.loads(path.read_text(encoding='utf-8'))
print(type(data).__name__, type(data['samples']).__name__)
row = data['samples'][0]
print(row)
print('动作名：', data['classes'][row['label']])
```

## 把重复逻辑封装成函数


```python
def count_label(samples, label):
    count = 0
    for row in samples:
        if row['label'] == label:
            count += 1
    return count
print('类别0的训练样本数：', count_label(data['samples'], 0))
```

## 故意制造并理解异常，不让整个课中断


```python
try:
    print(row['不存在的键'])
except KeyError as error:
    print('KeyError表示字典中没有这个键：', error)
print('先看traceback最后一行，再看它指向的代码；不要直接忽略所有异常。')
```

## 动手练习

统计“喝水”的训练片段数。类别编号来自classes映射，先用classes.index查找drinking的实际名字。可先打印classes。

显示“待完成”表示练习尚未作答；代码示例运行成功不等于练习完成。


```python
print(list(enumerate(data['classes'])))
your_answer = count_label(data['samples'], 6)  # TODO：填入统计结果，勿猜类别编号
if your_answer is None:
    print('待完成：在清单中定位drink对应的类，再调用count_label')
else:
    drink_id = next(i for i, n in enumerate(data['classes']) if 'drink' in n.lower() and 'pour' not in n.lower() and 'stir' not in n.lower())
    assert your_answer == sum(r['label'] == drink_id for r in data['samples'])
    print('练习通过')
```
