用numel统计tiny_video_model的全部参数数量，填入your_answer。


```python
import torch
from torch import nn
model = nn.Sequential(nn.Conv3d(4, 8, 3, padding=1), nn.ReLU(), nn.AdaptiveAvgPool3d(1), nn.Flatten(), nn.Linear(8, 40))
your_answer = sum(p.numel() for p in model.parameters())
assert your_answer == (8*4*3*3*3 + 8) + (40*8 + 40)
print(your_answer)
```

## 为什么这样写

Conv3d参数=输出通道×输入通道×核体积+偏置；线性层=输出×输入+偏置，共1232。冻结参数仍是模型参数，不能从总量中忽略。

解释空间与时间信息；预训练/微调/最终权重区别；说出4通道来源。
