从cm自行计算百分制accuracy与Macro F1，缺少支持/预测的类别F1按0，始终平均40类。填入包含accuracy和macro_f1的字典。


```python
import numpy as np
from analyze_predictions import analyze
report = analyze(L.PROJECT / L.CONFIG['reference_run'] / 'validation.csv', L.PROJECT / 'data/raw/validation/manifest.json')
cm = np.asarray(report['confusion_matrix'])
tp = cm.diagonal()
f1 = 2 * tp / np.maximum(1, cm.sum(axis=0) + cm.sum(axis=1))
your_answer = {'accuracy': 100*tp.sum()/cm.sum(), 'macro_f1': 100*f1.mean()}
assert abs(your_answer['accuracy'] - report['metrics']['accuracy']) < 1e-6
assert abs(your_answer['macro_f1'] - report['metrics']['macro_f1']) < 1e-6
print(your_answer)
```

## 为什么这样写

对角线是预测正确的样本数；Macro F1不按样本数加权。分母为0时TP也为0，置分母1得到F1=0。

独立计算两个指标；三组错例有真实片段；区分推测原因与已经验证的原因。
