统计“喝水”的训练片段数。类别编号来自classes映射，先用classes.index查找drinking的实际名字。可先打印classes。


```python
data = L.manifest('train')
drink_id = next(i for i, n in enumerate(data['classes']) if 'drink' in n.lower() and 'pour' not in n.lower() and 'stir' not in n.lower())
your_answer = sum(row['label'] == drink_id for row in data['samples'])
assert your_answer > 0
print(drink_id, data['classes'][drink_id], your_answer)
```

## 为什么这样写

sum可以统计布尔条件为True的次数。实际类名必须来自清单；英文名字的查找只用于本地入门练习，正式类别映射沿用完整classes数组。
