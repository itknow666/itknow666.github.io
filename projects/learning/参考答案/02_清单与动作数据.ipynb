查找训练集中第一段喝水片段，并输出它在samples中的位置，填入your_answer。选读：修改L.preview的index查看它。


```python
train = L.manifest('train')
drink_id = next(i for i, n in enumerate(train['classes']) if 'drink' in n.lower() and 'pour' not in n.lower() and 'stir' not in n.lower())
your_answer = next(i for i, row in enumerate(train['samples']) if row['label'] == drink_id)
assert train['samples'][your_answer]['label'] == drink_id
from IPython.display import display, Image
display(Image(filename=str(L.preview('train', your_answer))))
print('样本位置', your_answer, '类别', drink_id)
```
