把batch_size改为3。四个样本且drop_last=False时应产生几个batch、大小各是多少？填入列表your_answer。


```python
import torch
from torch.utils.data import DataLoader, TensorDataset
dataset = TensorDataset(torch.arange(4))
your_answer = [len(batch[0]) for batch in DataLoader(dataset, batch_size=3, drop_last=False)]
assert your_answer == [3, 1]
print(your_answer)
```

## 为什么这样写

4=3+1，保留最后batch避免遗漏样本；它的batch维可能与前面不同。

解释Dataset/DataLoader分工；指出B/C/T/H/W；说清num_workers=0用途。
