从scores找分数最高的三个类别编号，按分数从高到低排列。填入your_answer。


```python
import numpy as np
model, _ = L.load_reference_model()
path, _, classes = L.sample_path('validation', 0)
scores = np.asarray(model.predict(path))
your_answer = np.argsort(scores)[::-1][:3]
assert len(set(your_answer.tolist())) == 3
assert scores[your_answer[0]] >= scores[your_answer[1]] >= scores[your_answer[2]]
print([(int(i), classes[i]) for i in your_answer])
del model
```

## 为什么这样写

argsort返回下标；[::-1]反转升序。top3仅说明相对分数，未经校准不能当作置信保证。

能从新终端复现预测；解释Model/config/weights_path；理解score、class_id、class_name。
