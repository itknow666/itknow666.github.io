```python
from pathlib import Path
import sys
sys.dont_write_bytecode = True
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents]
            if (p / 'tools/learning.py').is_file() and (p / 'config.json').is_file())
sys.path.insert(0, str(ROOT / 'tools'))
import learning as L
OUT = ROOT / 'outputs'
OUT.mkdir(exist_ok=True)
```

比较两个config，返回值不同的键的集合，填入your_answer。运行目录名称不同，但不是模型超参数。


```python
a = L.train('answer-plan-a', 5e-5, dry_run=True)
b = L.train('answer-plan-b', 1e-5, dry_run=True)
your_answer = {key for key in a['config'] if a['config'][key] != b['config'][key]}
assert your_answer == {'lr'}
assert not Path(a['output']).exists() and not Path(b['output']).exists()
print(your_answer)
```

## 为什么这样写

比较配置时不能顺手改batch或epoch。seed相同降低部分随机差异，但CUDA并不保证逐位确定性，多种子验证属于后续研究。

说清唯一变量；找到summary/history/progress；写出证据和局限。
