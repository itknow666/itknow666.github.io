设计一个损坏CSV：包含重复sample_id，并检查校验器确实拒绝。填入your_answer=True表示捕获到预期错误。


```python
import csv
path = OUT / 'invalid-example.csv'
with path.open('w', encoding='utf-8', newline='') as stream:
    writer = csv.writer(stream)
    writer.writerow(['sample_id','predicted_class'])
    writer.writerows([['duplicate',0],['duplicate',1]])
your_answer = False
try:
    L.validate_csv(path)
except ValueError as error:
    your_answer = True
    print('按预期拒绝：', error)
assert your_answer is True
```

## 为什么这样写

校验器不仅检查行数，还检查列名、ID集合、重复和类别范围。故意损坏的文件仅放在学习区outputs中。
